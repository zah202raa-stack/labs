```python
# 6.2 Cleaning Our Data
# A. Write a profit_margin function that accepts a row of data, which is a dictionary
#    It should return the result of dividing the profit column by the sales column (i.e. profit/sales)
def profit_margin(x) :
    return x['profit']/x['sales']

```

```python
# B. Create a new column in the orders dataframe called 'profit_margin' by applying the profit margin function row-by-row
# Your code here please
orders['profit_margin']=orders.apply(profit_margin, axis=1)
orders.head(2)
```

We also want to create a column called `margin_categorization` that categorizes our new profit margin as "profitable", "unprofitable", or "break even". We can create a new function to do this, and then apply it to our orders DataFrame using the same method as above.

The function should follow these rules:
- If the profit_margin is greater than 0, the function should return "profitable"
- If the profit_margin is 0, the function should return "break even"
- If the profit_margin is less than 0, the function should return "unprofitable"

```python
# C. Use the same process to create a new column called margin_category
# If the profit_margin is less than 0, the margin_category should be "unprofitable"
# If the profit_margin is 0, the margin_category should be "break even"
# If the profit_margin is above 0, the margin_category should "profitable"
def margin_categorization(row):
    if row<0 :
        return 'unprofitable'
    elif row==0 :
      return'break even'
    elif row>0 :
        return'profitable'
# continue your code here

orders['margin_category']=orders['profit_margin'].apply(margin_categorization)
    
# apply the function to the orders dataframe 
# to create the new column 'margin_category'
orders.head()
```
